## Introdução



O módulo `random` é um módulo embutido de Python utilizado para operações de sorteios (pseudo)aleatórios.



```python
import random
```

## Sortear números



### Sortear um número real dentro do intervalo $[0,1[$



Use o `random.random` sem passar nenhum argumento.



```python
real_aleatorio_entre_0_e_1 = random.random()

real_aleatorio_entre_0_e_1
```

### Sortear um número inteiro dentro do intervalo $[a,b]$



Use o `random.randint` e passe como argumento os valores $a$ e $b$.



```python
a = 3
b = 10

inteiro_aleatorio_entre_a_e_b = random.randint(a,b)


inteiro_aleatorio_entre_a_e_b
```

**Nota**: existe também o `randrange` que segue a mesma ideia do `randint`, porém o intervalo de sorteio é $[a,b[$ (isto é, aberto em $b$, similar ao que ocorre com o `range` do Python).



```python
inteiro_aleatorio_entre_a_e_b_com_b_aberto = random.randrange(a,b)

inteiro_aleatorio_entre_a_e_b_com_b_aberto
```

### Sortear um número real dentro do intervalo $[a,b]$



Use o `random.uniform` e passe como argumento os valores $a$ e $b$.



```python
a = 1.25
b = 4.73

real_aleatorio_entre_a_e_b = random.uniform(a, b)

real_aleatorio_entre_a_e_b
```

### Sortear um número real a partir de uma distribuição normal (Gaussiana)



Use o `random.gauss` e passe a média e o desvio padrão da sua distribuição normal como argumentos.

**Nota**: sortear valores de uma distribuição normal é bastante útil para quem está cursando a disciplina de Probabilidade e Estatística.



```python
media = 0
desvio_padrao = 1

real_sorteado_de_uma_distribuicao_normal = random.gauss(media, desvio_padrao)

print(real_sorteado_de_uma_distribuicao_normal)
```

## Sortear itens



### Tenho uma lista e quero sortear um item



Para isso use o `random.choice`.



```python
lista = 'abcdefg'

escolha_aleatoria = random.choice(lista)

escolha_aleatoria
```

Observe que sortear um item de uma lista não altera a lista original.



```python
lista
```

### Tenho uma lista e quero sortear $n$ itens com reposição



Para isso use o `random.choices` e passe o valor de $n$ para o argumento `k`. Note que é &ldquo;choices&rdquo; no plural, no anterior era &ldquo;choice&rdquo; no singular. Note também que este é um sorteio *com reposição*, isto é: um valor sorteado pode ser sorteado novamente.



```python
n = 3

lista = [
    "azul",
    "vermelho",
    "roxo",
    "amarelo",
    "marrom",
    "preto",
    "branco",
    "verde",
]

escolhas_aleatorias = random.choices(lista, k=n)

escolhas_aleatorias
```

### Tenho uma lista e quero sortear $n$ itens sem reposição



Use o `random.sample` para realizar sorteios sem reposição (isto é, um item sorteado não pode ser sorteado novamente). O argumento `k` controla quantos itens serão sorteados.



```python
n = 3

lista = [
    "azul",
    "vermelho",
    "roxo",
    "amarelo",
    "marrom",
    "preto",
    "branco",
    "verde",
]

sorteio_aleatorio_sem_reposicao = random.sample(lista, k=n)

sorteio_aleatorio_sem_reposicao
```

### Tenho uma lista e quero sortear um item, mas a chance de sortear cada item não é igual



Novamente usaremos o `random.choices` aqui. Precisamos passar a chance de sortear cada item no argumento `weights` (que significa &ldquo;pesos&rdquo; em inglês). Veja que o argumento `weights` deve receber uma lista de mesmo tamanho da lista do sorteio contendo a chance de cada item da lista ser sorteado. **As chances não precisam somar 100%**, o Python fará essa conta pra você. Quanto maior o peso, maior a chance do respectivo item ser sorteado. No exemplo abaixo, as cores &ldquo;roxo&rdquo; e &ldquo;verde&rdquo; são as mais prováveis de serem sorteadas (peso 5), já a cor amarelo é a que tem a menor chance de ser sorteada (peso 0.2).



```python
lista = [
    "azul",
    "vermelho",
    "roxo",
    "amarelo",
    "marrom",
    "preto",
    "branco",
    "verde",
]

chance = [
    1,
    3,
    5,
    0.2,
    1,
    3,
    3,
    5,
]

escolha_aleatoria_ponderada = random.choices(lista, weights=chance)

print(escolha_aleatoria_ponderada)
```

### Tenho uma lista e quero sortear $n$ itens com reposição, mas a chance de sortear cada item não é igual



Basicamente a mesma coisa que fizemos acima. Veja o código abaixo e verá que não tem nada de novo aqui.



```python
n = 3

lista = [
    "azul",
    "vermelho",
    "roxo",
    "amarelo",
    "marrom",
    "preto",
    "branco",
    "verde",
]

chance = [
    10,
    10,
    2,
    0.2,
    1,
    3,
    10,
    2,
]

escolhas_aleatorias_ponderadas = random.choices(lista, weights=chance, k=3)

print(escolhas_aleatorias_ponderadas)
```

## Embaralhar uma lista de objetos



Você pode fazer isso de duas maneiras diferentes. Se quiser embaralhar a própria lista, efetivamente alterando o objeto em si, use o `random.shuffle`.



```python
lista = [
    "azul",
    "vermelho",
    "roxo",
    "amarelo",
    "marrom",
    "preto",
    "branco",
    "verde",
]

random.shuffle(lista)

lista
```

Um erro clássico de que está aprendendo Python está apresentado na célula abaixo. Você consegue ver qual é o problema aqui?



```python
lista = random.shuffle(lista)

print(lista)
```

Se você não quiser alterar a lista original, mas sim quer gerar um objeto novo com os elementos da lista original embaralhados, então use o `random.sample` com um valor de `k=len(objeto)`.



```python
lista = [
    "azul",
    "vermelho",
    "roxo",
    "amarelo",
    "marrom",
    "preto",
    "branco",
    "verde",
]

lista_embaralhada = random.sample(lista, k=len(lista))

lista_embaralhada
```

## Como um computador pode gerar números aleatórios?



Na verdade, o módulo `random` não gera números aleatórios pois o computador não tem um mecanismo interno que seja capaz de gerar números aleatórios.

Por isso, o gerador de números aleatórios do Python é chamado de **pseudoaleatório** porque, na prática, os números são gerados por um **algoritmo determinístico** a partir de um valor inicial chamado semente (*seed*, em inglês). Isso significa que, fixando a semente, os resultados são reprodutíveis. Tente rodar a célula abaixo quantas vezes desejar, seu resultado será sempre o mesmo.



```python
random.seed(42)

print(random.randint(1, 100))
```

### Simulando um dado



Crie uma função que recebe um inteiro positivo $n$. Esta função irá jogar um dado de seis faces $n$ vezes e retornar para o usuário a soma de todas as jogadas.



```python
import random

def soma_dado(n):

    faces = []

    i=1
    
    while i<=n:

        face = random.randint(1,6)
        faces.append(face)
        i+=1

    return sum(faces), faces

soma, lista = soma_dado(2)

print(soma)
lista
```

### A senha aleatória



Faça uma função que recebe um inteiro positivo $n$. Esta função irá gerar uma senha aleatória de tamanho $n$ usando letras maiúsculas, letras minúsculas e dígitos.



```python
minusculas = "abcdefghijklmnopqrstuvwxyz"
maiusculas = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
digitos = "0123456789"

lista = [minusculas, maiusculas, digitos]

def senha_aleatoria(n):
    """Retorna uma senha aleatória com n dígitos com números e letras maiúsculas ou minúsculas"""

    #primeiro uma variável que receberá os caracteres
    senha_lista = []
    i = 0
    
    while i<n:
        char = random.choice(random.choice(lista))
        senha_lista.append(char)
        i+=1

    senha = "".join(senha_lista)

    return senha

senha_aleatoria(7)
```

### Dez caras



Faça uma função que joga uma moeda dez vezes e conta o número de caras. Se caírem 10 caras, a função para. Do contrário, a função joga a moeda dez vezes novamente. O retorno da função é o número de tentativas até que ela obtenha 10 caras.



```python
def cara_coroa():

    probabilidade = ['cara','coroa']

    lista = []

    contador = 0

    while lista.count('cara')<=10:

        i = 1
        while i<=10:
            lista.append(random.choice(probabilidade))
            i+=1
        contador+=1

    return contador

cara_coroa()
```

**Extra**: qual é o valor médio do número de tentativas se nós executarmos a função anterior dez mil vezes?



```python
def cara_coroa():

    probabilidade = ['cara','coroa']

    tentativas = []

    for i in range(0,10000):
        
        contador = 0
        lista = []

        while lista.count('cara')<=10:
    
            i = 1
            while i<=10:
                lista.append(random.choice(probabilidade))
                i+=1
            contador+=1

        tentativas.append(contador)
        

    return sum(tentativas)/len(tentativas)

cara_coroa()
```

### O jogo da adivinhação



Crie uma função que sorteia um número inteiro no intervalo $[0, 100]$. Dentro da função, o usuário é convidado a adivinhar o número sorteado (use o comando `input`). Para cada palpite do usuário existem três possibilidades:

1.  O palpite estava certo! A função é encerrada e retorna ao usuário o número de tentativas.
2.  O palpite é menor que o valor sorteado. A função printa &ldquo;Seu palpite é menor que o valor que eu sorteei, tente novamente!&rdquo; e pede outro palpite.
3.  O palpite é maior que o valor sorteado. A função printa &ldquo;Seu palpite é maior que o valor que eu sorteei, tente novamente!&rdquo; e pede outro palpite.



```python
import random

def sorteio():
    """O usuário escreve o palpite de um número de 0 a 100 e caso acerte a função acaba"""
    lista = list(range(101))

    numero = random.choice(lista)

    contador = 0
    
    while True:
        palpite = int(input("Digite seu palpite: "))
        if palpite==numero:
            contador+=1
            return f"O palpite estava certo! Acertou em {contador} tentativas!"
        elif palpite<numero:
            print("Seu palpite é menor que o valor que eu sorteei, tente novamente!")
            contador+=1
        else:
            print("Seu palpite é maior que o valor que eu sorteei, tente novamente!")
            contador+=1

sorteio()
    
```
